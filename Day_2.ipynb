```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("hr_dataset.csv")
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
#Work on Employee ID
df["employee_id"] = df["employee_id"].str.replace("-","")
df["employee_id"] = df["employee_id"].str.replace("_","")
df["employee_id"] = df["employee_id"].str.replace("ID","E")
df["employee_id"] = df["employee_id"].str.replace("EMP","E")

#Name Columns
df["name"] = df["name"].str.replace("##","")
df["name"] = df["name"].str.replace("12","")
df["name"] = df["name"].str.replace("@","")
df["name"] = df["name"].str.lower().str.capitalize()

#Age Columns
df["age"] = df["age"].abs()
df["age"] = df["age"].where(df["age"]<100 ,np.nan)
df["age"] = df["age"].fillna(df["age"].median()).astype(int)

#Department Columns
df["department"] = df["department"].str.replace(" ","")
df["department"] = df["department"].str.replace(".","")
df["department"] = df["department"].str.lower().str.upper()
df["department"] = df["department"].fillna(df["department"].mode()[0]) 
#Salary 
df["salary"] = df["salary"].str.replace("k","000")
df["salary"] = df["salary"].str.replace("-","")
df["salary"] = df["salary"].astype(float)
df["salary"] = df["salary"].fillna(df["salary"].mean()).astype(int)


#Email Columns
df["email"] = df["email"].astype(str)
df["email"] = df["email"].replace("gamail","gmail")
df["email"] = df["email"].str.replace("noemail","")
df["email"] = df["email"].replace("invalid@","")

# df.loc[rows,columns]
df.loc[df["email"].str.endswith("@"),"email"] = (df["email"] +  "gmail.com")

df["email"] = df["email"].replace("nan","")
df["email"] = df["email"].replace("","Not Provided")


#Join Date
df["join_date"] = df["join_date"].str.replace("/","-")
df["join_date"] = df["join_date"].fillna("Not Provided")

#Experiance Columns
df["experience_years"] = df["experience_years"].abs()
df["experience_years"] = df["experience_years"].fillna(df["experience_years"].mean()).astype(int)

#City Columns.
df["city"] = df["city"].str.lower().str.capitalize()
df["city"] = df["city"].replace("Suratt","Surat")
df["city"] = df["city"].fillna(df["city"].mode()[0])


#Performance
df["performance"] = df["performance"].str.lower().str.capitalize()
df["performance"] = df["performance"].fillna("Not Provided")

df["city"].mode()
# df.to_csv("HR_Cleaned_Dataset.csv")
```
